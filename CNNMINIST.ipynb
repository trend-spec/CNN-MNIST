```python
# ============================================
# CNN IMAGE CLASSIFICATION ON FASHION-MNIST
# KERAS + TENSORFLOW
# ============================================

# Step 2: Import the libraries

import keras
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt

from keras.models import Sequential
from keras.layers import Dense, Dropout, Flatten
from keras.layers import Conv2D, MaxPooling2D
from keras.utils import to_categorical

from sklearn.model_selection import train_test_split


# ============================================
# Step 3: Load Fashion-MNIST dataset
# ============================================

from tensorflow.keras.datasets import fashion_mnist

(X, y), (test, test_y) = fashion_mnist.load_data()

print("Training data shape:", X.shape)
print("Testing data shape:", test.shape)


# ============================================
# Pre-processing
# ============================================

# Add channel dimension

X = X.reshape(X.shape[0], 28, 28, 1)
test = test.reshape(test.shape[0], 28, 28, 1)

# Normalize pixel values

X = X.astype("float32") / 255.0
test = test.astype("float32") / 255.0

# Convert labels to one-hot encoding

y = to_categorical(y, 10)
test_y = to_categorical(test_y, 10)


# ============================================
# Step 4: Create validation set
# ============================================

X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    random_state=42,
    test_size=0.2
)

print("Training:", X_train.shape)
print("Validation:", X_test.shape)


# ============================================
# Step 5: Define the model structure
# ============================================

model = Sequential()

# First convolution layer

model.add(
    Conv2D(
        32,
        kernel_size=(3, 3),
        activation="relu",
        input_shape=(28, 28, 1)
    )
)

model.add(MaxPooling2D(pool_size=(2, 2)))


# Second convolution layer

model.add(
    Conv2D(
        64,
        kernel_size=(3, 3),
        activation="relu"
    )
)

model.add(MaxPooling2D(pool_size=(2, 2)))


# Flatten layer

model.add(Flatten())


# Dense hidden layer

model.add(Dense(128, activation="relu"))


# Output layer

model.add(Dense(10, activation="softmax"))


# ============================================
# Compile the model
# ============================================

model.compile(
    loss="categorical_crossentropy",
    optimizer="adam",
    metrics=["accuracy"]
)

model.summary()


# ============================================
# Step 6: Train the model
# ============================================

model.fit(
    X_train,
    y_train,
    epochs=10,
    validation_data=(X_test, y_test)
)


# ============================================
# Step 7: Making predictions
# ============================================

prediction = model.predict(test)

prediction = np.argmax(prediction, axis=1)

print("Predicted classes:")
print(prediction[:20])
```
